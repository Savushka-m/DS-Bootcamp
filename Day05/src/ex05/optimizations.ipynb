```python
import pandas as pd
import gc
```

### Считаем файл fines.csv полученный в прошлом задании

```python
df = pd.read_csv('../data/fines.csv')
```

### Иттерации: вычислим штрафы/возврат*год для каждой строки и создадим новые столбцы с вычисленными данными

```python
def loop(df):
    sp = []
    for i in range(len(df)):
        t = df.iloc[i]['Fines'] / df.iloc[i]['Refund'] * df.iloc[i]['Year']
        sp.append(t)
    df['Сalculated'] = sp
```

```python
%%timeit
loop(df)
```

```python
def iter(df):
    sp = []
    for index, row in df.iterrows():
        t = row['Fines'] / row['Refund'] * row['Year']
        sp.append(t)
    df['Сalculated'] = sp
```

```python
%%timeit
iter(df)
```

```python
def appl(df):
    df['Сalculated'] = df.apply(lambda x: x['Fines'] / x['Refund'] * x['Year'], axis = 1)
```

```python
%%timeit
appl(df)
```

```python
def series(df):
    df['Сalculated'] = df['Fines'] / df['Refund'] * df['Year']
```

```python
%%timeit
series(df)
```

```python
def series_values(df):
    df['Сalculated'] = df['Fines'].values / df['Refund'].values * df['Year'].values
```

```python
%%timeit
series_values(df)
```
