```python
import pandas as pd
```

### Считаем файл в датафрейм

```python
df = pd.read_csv('../data/feed-views.log',
                 sep='\t',
                 skiprows=[2, 3],
                 skipfooter=2,
                 names = ['datetime', 'user'],
                 index_col='datetime',
                 engine='python')
```

### Переименуем datetime в date_time

```python
df.index.name = 'date_time'
```

### Сохраним в формате csv, в качестве разделителя воспользуемся ';'

```python
df.to_csv('../data/feed-views-semicolon.log', sep=';')
```
