```python
import pandas as pd
```

### Считаем файл в датафрейм и зададим индекс

```python
df = pd.read_csv('../data/auto.csv',
                  sep=',',
                  index_col='ID',
                  engine='python')
```

### Подсчитаем кол-во наблюдений

```python
df.count()
```

### Удалим дубликаты принимая во внимание только столбцы CarNumber, Make_n_model, Fines

```python
df.drop_duplicates(subset=['CarNumber', 'Make_n_model', 'Fines'],
                        keep='last',
                        inplace=True)
```

```python
df.count()
```

### Теперь займемся пропущенными значениями

```python
df.isnull().sum()
```

```python
df.dropna(axis='columns',
          thresh=500, 
          inplace=True)
df.isnull().sum()
```

```python
df['Refund'].fillna(method='ffill', inplace=True)
df.isnull().sum()
```

```python
df['Fines'].fillna(df['Fines'].mean(), inplace=True)
df.isnull().sum()
```

### Проанализируем марку и модель

```python
df['Make'] = df['Make_n_model'].apply(lambda x: x.split()[0])
df['Model'] = df['Make_n_model'].apply(lambda x: x.split()[1] if len(x.split()) > 1 else None)
df.drop(labels='Make_n_model',
        axis='columns', 
        inplace=True)
```

```python
df.to_json('../data/auto.json', orient='records')
```
