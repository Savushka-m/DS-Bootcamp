```python
import pandas as pd
```

### Создадим датафрейм со столбцами datetime и user, с последующим добавлением столбцов year, month и т.д.

```python
df = pd.read_csv('../data/feed-views.log', 
                 sep='\t', 
                 names = ['datetime', 'user'], 
                 engine='python')
df['datetime'] = pd.to_datetime(df['datetime'])
df['year'] = df['datetime'].dt.year
df['month'] = df['datetime'].dt.month
df['day'] = df['datetime'].dt.day
df['hour'] = df['datetime'].dt.hour
df['minute'] = df['datetime'].dt.minute
df['second'] = df['datetime'].dt.second
```

```python
df.head()
```

### Создадим столбец daytime и установим столбец user индексом

```python
df['daytime'] = pd.cut(df['hour'],
                       bins=[0, 4, 7, 11, 17, 20, 24],
                       labels=['night', 'early morning', 'morning', 'afternoon', 'early evening', 'evening'],
                       right=False)
df.set_index('user', inplace=True)
```

```python
df.head()
```

### Посчитаем кол-во элементов в датафрейме

```python
df.count()
```

```python
df['daytime'].value_counts()
```

### Отсортируем значения по столбцам hour, minute, second в порядке возрастания

```python
df.sort_values(['hour','minute', 'second'])
```

### Рассчитаем различные минимумы и максимумы 

```python
night_df = df[df['daytime'] == 'night']
max_hour_night = night_df['hour'].max()
print(max_hour_night)
```

```python
morning_df = df[df['daytime'] == 'morning']
min_hour_morning = morning_df['hour'].min()
print(min_hour_morning)
```

```python
df[df['hour'] == max_hour_night]
```

```python
df[df['hour'] == min_hour_morning]
```

```python
df['daytime'].mode()
```

### Выведем 3 самых ранних часа и 3 самых поздних часа в категории morning

```python
morning_df.nsmallest(3, 'hour')
```

```python
morning_df.nlargest(3, 'hour')
```
